# Chạy demo đồ án

```python
import subprocess
import os
import sys
import time

# Xác định chính xác thư mục gốc dự án
goc_du_an = os.path.abspath(os.getcwd())

# Đảm bảo terminal output hỗ trợ tiếng Việt có dấu
if sys.platform == "win32":
    os.system('chcp 65001 > nul')

print("=" * 60)
print("KÍCH HOẠT DỰ ÁN PHÂN LOẠI RÁC THẢI")
print("=" * 60)

# 1. Bật Backend Service (port 5000)
path_backend = os.path.join(goc_du_an, "Backend", "backend_service.py")
print("1. Đang khởi động Backend Service...")
process_backend = subprocess.Popen([sys.executable, path_backend])
time.sleep(2) 

# 2. Bật AI Service (port 8000)
path_ai = os.path.join(goc_du_an, "AI_Service", "ai_service.py")
print("2. Đang khởi động AI Service...")
process_ai = subprocess.Popen([sys.executable, path_ai])
time.sleep(3) 

path_giao_dien_moi = os.path.join(goc_du_an, "UI", "main_app.py")
print("3. Đang mở Giao diện chính...")
print("-" * 60)

try:
    # Cấu hình giải phóng luồng hiển thị đồ họa và bật log lỗi nếu sập
    result = subprocess.run(
        [sys.executable, path_giao_dien_moi],
        capture_output=True,  # Bắt lại để tóm gọn lỗi nội bộ của main_app.py nếu có
        text=True,
        encoding='utf-8',
        errors='ignore'
    )
    
    if result.returncode != 0:
        print("\n[LỖI NỘI BỘ MAIN_APP.PY] Lỗi! Nguyên nhân:")
        print("-" * 50)
        print(result.stderr)
        print("-" * 50)
    else:
        print("\n[THÔNG BÁO] Đóng giao diện chính.")

except KeyboardInterrupt:
    print("\n[THÔNG BÁO] Ngắt kết nối từ bàn phím.")

finally:
    # Dọn dẹp cổng kết nối an toàn khi kết thúc
    print("\n" + "=" * 60)
    print("Đang đóng các dịch vụ chạy ngầm để giải phóng bộ nhớ...")
    try:
        process_backend.terminate()
        print("[ĐÃ TẮT] Đã đóng cổng Backend")
    except:
        pass
    try:
        process_ai.terminate()
        print("[ĐÃ TẮT] Đã đóng cổng AI Service")
    except:
        pass
    print("KẾT THÚC HỆ THỐNG PHÂN LOẠI RÁC THẢI.")
    print("=" * 60)
```
