# Chạy thử chương trình

### File code để chạy các quy trình của đồ án

```python
import subprocess
import os
import sys
import time

# Xác định đường dẫn thư mục gốc của dự án
goc_du_an = os.getcwd()

print("--- HỆ THỐNG KÍCH HOẠT DỰ ÁN TỰ ĐỘNG ---")

# 1. Bật Backend Service (Chạy ngầm ở cổng 5000)
path_backend = os.path.join(goc_du_an, "Backend", "backend_service.py")
print("⚡ Đang khởi động Backend Service...")
process_backend = subprocess.Popen([sys.executable, path_backend])
time.sleep(2) # Đợi 2 giây cho Backend mở cổng ổn định

# 2. Bật AI Service (Chạy ngầm ở cổng 8000)
path_ai = os.path.join(goc_du_an, "AI_Service", "ai_service.py")
print("🧠 Đang khởi động AI Service...")
process_ai = subprocess.Popen([sys.executable, path_ai])
time.sleep(3) # Đợi 3 giây cho AI tải xong Model ONNX

# 3. Bật Giao diện Camera chính
path_camera = os.path.join(goc_du_an, "CameraService", "main.py")
print("📸 Đang bật Giao diện Camera...")

try:
    # Chạy và lấy toàn bộ log lỗi nếu có từ file main.py
    result = subprocess.run(
        [sys.executable, path_camera],
        capture_output=True,
        text=True,
        encoding='utf-8',
        errors='ignore'
    )
    
    # Nếu file main.py bị lỗi (exit code khác 0), in lỗi ra màn hình
    if result.returncode != 0:
        print("\n❌ [LỖI NỘI BỘ MAIN.PY] Chi tiết lỗi khiến file camera bị sập:")
        print(result.stderr)
    else:
        print("\n[THÔNG BÁO] Bạn đã chủ động tắt giao diện camera.")

except KeyboardInterrupt:
    print("\n[THÔNG BÁO] Người dùng ngắt hệ thống từ bàn phím.")

finally:
    # Khi bạn tắt cửa sổ Camera Tkinter hoặc tắt Notebook, lệnh này mới được chạy để dọn dẹp
    print("\n⚠️ Đang dọn dẹp và đóng các dịch vụ chạy ngầm...")
    process_backend.terminate()
    process_ai.terminate()
    print("[ĐÃ TẮT] Hệ thống đã đóng an toàn giải phóng bộ nhớ.")
```
