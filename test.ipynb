# Chạy demo đồ án

```python
import subprocess
import os
import sys
import time

# Xac dinh chinh xac duong dan goc du an
goc_du_an = os.path.abspath(os.getcwd())

print("=" * 60)
print("HE THONG KICH HOAT DU AN TU DONG KHEP KIN")
print("=" * 60)
print(f"Thu muc dang dung: {goc_du_an}\n")

# 1. Bat Backend Service (Cong 5000)
path_backend = os.path.join(goc_du_an, "Backend", "backend_service.py")
print("1. Dang khoi dong Backend Service...")
process_backend = subprocess.Popen([sys.executable, path_backend])
time.sleep(2) 

# 2. Bat AI Service (Cong 8000)
path_ai = os.path.join(goc_du_an, "AI_Service", "ai_service.py")
print("2. Dang khoi dong AI Service...")
process_ai = subprocess.Popen([sys.executable, path_ai])
time.sleep(3) 

# 3. Bat Giao dien Camera chinh (Da cap nhat sang thu muc UI va file TempInterface.py)
path_camera = os.path.join(goc_du_an, "UI", "TempInterface.py")
print("3. Dang mo Giao dien Camera chinh...")
print("-" * 60)

try:
    # Chay file giao dien truc tiep de tranh nghen luong do hoa CustomTkinter
    result = subprocess.run(
        [sys.executable, path_camera],
        text=True,
        encoding='utf-8',
        errors='ignore'
    )
    
    if result.returncode != 0:
        print("\n[LOI NOI BO TEMPINTERFACE.PY] File giao dien bi sap!")
    else:
        print("\n[THONG BAO] Ban da chu dong dong Giao dien chinh.")

except KeyboardInterrupt:
    print("\n[THONG BAO] Nguoi dung ngat tu ban phim.")

finally:
    # Don dep cong ket noi an toan khi ket thuc
    print("\n" + "=" * 60)
    print("Dang dong cac dich vu chay ngam giai phong bo nho...")
    try:
        process_backend.terminate()
        process_backend.wait(timeout=2)
        print("[DA TAT] Da dong cong Backend (5000)")
    except:
        pass
    try:
        process_ai.terminate()
        process_ai.wait(timeout=2)
        print("[DA TAT] Da dong cong AI Service (8000)")
    except:
        pass
    print("HE THONG DA DUOC GIAI PHONG AN TOAN!")
    print("=" * 60)
```
